```python
def remove_pps_outliers(df, group_col, target_col):
    """
    Removes outliers from the target column based on mean and standard deviation,
    grouped by a specified column.

    Parameters:
        df (pd.DataFrame): The input DataFrame.
        group_col (str): The column to group by (e.g., 'Location').
        target_col (str): The column to filter outliers from (e.g., 'rate').

    Returns:
        pd.DataFrame: The DataFrame with outliers removed.
    """
    df_out = pd.DataFrame()
    
    for key, subdf in df.groupby(group_col):
        # Calculate mean and standard deviation for the target column
        mean = np.mean(subdf[target_col])
        std_dev = np.std(subdf[target_col])
        
        # Filter rows within one standard deviation of the mean
        filtered_df = subdf[
            (subdf[target_col] > (mean - std_dev)) &
            (subdf[target_col] <= (mean + std_dev))
        ]
        
        # Append the filtered data to the output DataFrame
        df_out = pd.concat([df_out, filtered_df], ignore_index=True)
    
    return df_out
```

```python
def plot_scatter(df5,Location):
    matplotlib.rcParams['figure.figsize'] = (15,10)
    mapo = {1: 'red', 2:'green', 3:'blue', 4:'yellow', 5:'pink', 6:'orange'}
    for i in mapo:
        bhkx = df5[(df5.Location == Location) & (df5.BHK == i)]
        plt.scatter(bhkx['area'], bhkx.Price, color = mapo[i], label = str(i)+'bhk', s=50)
    
    plt.xlabel('area')
    plt.ylabel('rate')
    plt.title(Location)
    plt.legend()
```

```python
def remove_iqr_outliers(df, bhk_col='BHK', area_col='area', rate_col='rate'):
    """
    Removes outliers from the dataset based on Total Area and Rate for each BHK category.

    Parameters:
        df (pd.DataFrame): The input DataFrame.
        bhk_col (str): Column name for BHK category (e.g., 'BHK').
        area_col (str): Column name for Total Area (e.g., 'Total Area').
        rate_col (str): Column name for Rate (e.g., 'Rate').

    Returns:
        pd.DataFrame: DataFrame with outliers removed.
    """
    df_cleaned = pd.DataFrame()

    # Process each BHK category separately
    for bhk in df[bhk_col].unique():
        bhk_df = df[df[bhk_col] == bhk]
        
        # Calculate IQR for Area
        Q1_area = bhk_df[area_col].quantile(0.25)
        Q3_area = bhk_df[area_col].quantile(0.75)
        IQR_area = Q3_area - Q1_area
        lower_bound_area = Q1_area - 1.5 * IQR_area
        upper_bound_area = Q3_area + 1.5 * IQR_area
        
        # Calculate IQR for Rate
        Q1_rate = bhk_df[rate_col].quantile(0.25)
        Q3_rate = bhk_df[rate_col].quantile(0.75)
        IQR_rate = Q3_rate - Q1_rate
        lower_bound_rate = Q1_rate - 1.5 * IQR_rate
        upper_bound_rate = Q3_rate + 1.5 * IQR_rate
        
        # Filter rows within bounds
        filtered_df = bhk_df[
            (bhk_df[area_col] >= lower_bound_area) & 
            (bhk_df[area_col] <= upper_bound_area) &
            (bhk_df[rate_col] >= lower_bound_rate) & 
            (bhk_df[rate_col] <= upper_bound_rate)
        ]
        
        # Append to cleaned dataset
        df_cleaned = pd.concat([df_cleaned, filtered_df], ignore_index=True)
    
    return df_cleaned
```

```python
def log_transform_data_safe(data, columns=None):
    """
    Apply log transformation to numeric columns safely.
    """
    if not isinstance(data, pd.DataFrame):
        raise ValueError("Input data must be a Pandas DataFrame.")
    
    transformed_data = data.copy()
    
    # Select numeric columns if no specific columns are provided
    if columns is None:
        columns = transformed_data.select_dtypes(include=[np.number]).columns
    
    for col in columns:
        if (transformed_data[col] <= 0).any():
            print(f"Skipping column '{col}' due to non-positive values.")
            continue
        transformed_data[col] = np.log1p(transformed_data[col])  # log(x + 1)
    
    return transformed_data
```

```python
import pandas as pd
import numpy as np

def clean_data(df, columns_to_transform):
    """
    Cleans the given DataFrame by:
    - Removing rows with non-positive values.
    - Capping outliers in certain columns.
    - Applying log transformation to specified numeric columns after handling non-positive values.
    
    Parameters:
    - df: The input pandas DataFrame to clean.
    - columns_to_transform: List of columns where log transformation should be applied.
    
    Returns:
    - df: The cleaned pandas DataFrame.
    """
    
    # 1. Handling Non-Positive Values: Removing rows with non-positive values in 'Price', 'Area', 'Rate'
    df = df[df['Price'] > 0]
    df = df[df['area'] > 0]
    df = df[df['rate'] > 0]

    # 2. Capping Outliers
    df['Price'] = df['Price'].clip(lower=100, upper=10000)
    df['BHK'] = df['BHK'].clip(lower=1, upper=10)

    # 3. Apply log transformation to specified columns (after handling non-positive values)
    for col in columns_to_transform:
        if col in df.columns:  # Check if the column exists
            if col in df.select_dtypes(include=[np.number]).columns:  # Apply only to numeric columns
                if (df[col] <= 0).any():
                    # Replace non-positive values with a small positive number (e.g., 1)
                    df[col] = df[col].apply(lambda x: x if x > 0 else 1)
                
                # Apply log transformation
                df[col] = np.log1p(df[col])  # log(x + 1) transformation
        else:
            print(f"Warning: Column '{col}' not found in DataFrame.")
    
    return df
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state = 10)
```

```python
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Ridge
from xgboost import XGBRegressor
base_learners = [('lr',LinearRegression()), ('xgb', XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)),('rf', RandomForestRegressor(n_estimators=100, random_state=42))]
meta_model = Ridge()
stacking_model = StackingRegressor(estimators = base_learners, final_estimator = meta_model)
stacking_model.fit(X_train,y_train)
stacking_model.score(X_test,y_test)
```

```python
stacking_model = XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
stacking_model.fit(X_train,y_train)
stacking_model.score(X_test,y_test)
```

```python
def predict_price(Location, area, Balcony, BHK):
    # Initialize an array for the features
    x = np.zeros(len(X.columns))  # Initialize an array of zeros for the input features
    x[0] = area
    x[1] = Balcony
    x[2] = BHK

    # Handle the location (assuming Location is one-hot encoded)
    loc_index = np.where(X.columns == Location)[0]  # Find the index of the location
    if len(loc_index) > 0:
        x[loc_index[0]] = 1  # Set the location index to 1 (one-hot encoding)
    else:
        print(f"Warning: Location '{Location}' not found in columns!")

    # Create a DataFrame with the input data
    input_data = pd.DataFrame([x], columns=X.columns)

    # Predict the log-transformed price using the stacking model
    predicted_log_price = stacking_model.predict(input_data)[0]
    print(f"Predicted Log(Price): {predicted_log_price}")  # Check the predicted log price

    # Reverse the log transformation (log1p -> expm1) to get the original price
    predicted_price = np.expm1(predicted_log_price)
    print(f"Predicted Price after Exponential Reversal: {predicted_price}")  # Check the reversed price

    return predicted_price

# Example usage
value = predict_price('Akshayanagar,Bangalore', 500, 1, 3)
print(f"Predicted Value: {value}")
```
